```python
# Import libraries for reading data and creating charts.

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Section One : Read the CSV file using pandas.

df = pd.read_csv("woocommerce-product-export.csv")
```

```python
# Display a concise summary of the DataFrame.
# This shows column names, non-null counts, and data types.

df.info()
```

```python
# Show summary statistics for numeric columns.

df.describe() 
```

```python
# Display the first five rows by default.

df.head()
```

```python
# Display the last five rows by default.

df.tail()
```

```python
# Print only the requested columns.

print(df[["total_profit", "month_number"]])
```

```python
# Display total profit by month as a bar chart.

plt.figure(figsize=(8, 5))

plt.bar(
    df["month_number"],
    df["total_profit"],
    color="tab:blue",
    width=0.8
)

plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

# Match the axis ticks shown in the example.
plt.xticks([2, 4, 6, 8, 10, 12])
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.ylim(0, 500000)

# Add dashed grid lines.
plt.grid(True, linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()  
```

```python
# Plot total profit for all months.


plt.figure(figsize=(8, 5))

plt.plot(
    df["month_number"],
    df["total_profit"],
    color="red",
    linestyle=":",
    linewidth=3,
    marker="o",
    markerfacecolor="black",
    markeredgecolor="black",
    label="Profit data of last year"
)

plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

plt.xticks(df["month_number"])
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.ylim(100000, 500000)

plt.legend(loc="lower right")

plt.tight_layout()
plt.show()
```

```python
# Print all product sales data
# and show monthly units sold for each product.


products = {
    "facecream": "Face cream Sales Data",
    "facewash": "Face wash Sales Data",
    "toothpaste": "Toothpaste Sales Data",
    "bathingsoap": "Bathingsoap Sales Data",
    "shampoo": "Shampoo Sales Data",
    "moisturizer": "Moisturizer Sales Data"
}

# Print monthly sales for all six products.
print(df[["month_number"] + list(products)].to_string(index=False))

# Create a separate line for each product.
plt.figure(figsize=(8, 6))

for column, label in products.items():
    plt.plot(
        df["month_number"],
        df[column],
        marker="o",
        label=label
    )

plt.xlabel("Month Number")
plt.ylabel("Sales units in number")
plt.title("Sales Data")

plt.xticks(df["month_number"])
plt.yticks([
    1000, 2000, 4000, 6000, 8000,
    10000, 12000, 15000, 18000
])
plt.ylim(500, 18000)
plt.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# Plot bathing soap sales for each month.


plt.figure(figsize=(8, 5))

plt.scatter(
    df["month_number"],
    df["bathingsoap"],
    color="tab:blue",
    label="bathing soap Sales data"
)

plt.xlabel("Month Number")
plt.ylabel("Number of units Sold")
plt.title("bathingsoap Sales data")

plt.xticks(df["month_number"])
plt.yticks([6000, 8000, 10000, 12000, 14000])

# Add dashed grid lines to match the example.
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# Section Two : Create a date versus temperature line chart.


date = ["25/12", "26/12", "27/12"]
temp = [8.5, 10.5, 6.8]

plt.figure(figsize=(8, 5))

plt.plot(date, temp)

plt.xlabel("Date")
plt.ylabel("temperature")
plt.title("Date-wise Temperature")

plt.tight_layout()
plt.show()
```

```python
# Plot the provided weight and height pairs.


height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]

plt.figure(figsize=(8, 5))

plt.plot(
    weight,
    height,
    color="green",
    linestyle="-.",
    marker="o",
    markersize=10,
    markerfacecolor="green",
    markeredgecolor="green",
    label="Height by weight"
)

plt.xlabel("Weight in kg.")
plt.ylabel("Height in cm.")
plt.title("Average weight with respect to average height.")
plt.legend(loc="lower right")

plt.tight_layout()
plt.show()
```

```python
# Additional visualization: Bubble chart of weight and height.

plt.figure(figsize=(9, 6))

bubbles = plt.scatter(
    weight,
    height,
    s=[(h - 100) * 8 for h in height],
    c=height,
    cmap="Greens",
    edgecolors="black",
    alpha=0.85
)

plt.colorbar(bubbles, label="Height in cm.")
plt.xlabel("Weight in kg.")
plt.ylabel("Height in cm.")
plt.title("Weight and Height — Bubble Chart")
plt.grid(True, linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```
